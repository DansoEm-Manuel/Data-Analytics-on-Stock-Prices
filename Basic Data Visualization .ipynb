```python
##importing data
import pandas as pd

df = pd.read_csv(r"C:\Users\Danso Emmanuel Effah\OneDrive\Desktop\DATA ANALYSIS\Projects\India Project\iris.csv")

###Clean column names
df.columns = df.columns.str.strip()


###checing collons 
print(df.head())
print(df.columns)
```

```python
##bar plot
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

sns.countplot(
    x=df.iloc[:, -1],
    palette='Set2'
)

plt.title("Count of Iris Species")
plt.xlabel("Species")
plt.ylabel("Count")

plt.show()
```

```python
## line chart
plt.figure(figsize=(10,5))

plt.plot(df.iloc[:, 0])

plt.title("Trend of Sepal Length")
plt.xlabel("Index")
plt.ylabel("Sepal Length")

plt.show()
```

```python
## scattered plot
plt.figure(figsize=(8,5))

sns.scatterplot(
    x=df.iloc[:, 0],
    y=df.iloc[:, 2],
    hue=df.iloc[:, -1],
    palette='deep'
)

plt.title("Sepal Length vs Petal Length")
plt.xlabel(df.columns[0])
plt.ylabel(df.columns[2])

plt.show()
```

```python
##exporting visuals as image
plt.figure(figsize=(8,5))

sns.scatterplot(
    x=df.iloc[:, 0],
    y=df.iloc[:, 2],
    hue=df.iloc[:, -1]
)

plt.title("Scatter Plot")

plt.savefig("scatter_plot.png")

plt.show()
```
